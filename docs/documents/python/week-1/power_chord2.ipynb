# Power Chord2 - A program to generate sound on your computer
## Author: Kartik Bulusu, Associate Research Professor, MAE Department, GWU
## Created: October 15, 2026
### Description: The Jupyter notebook is an extension of the power_chord.py program 
### written on Thonny. The output of the program is similar but has extra features such as 
### a function (def-command), if __name__ == "__main__", a for loop and an if-construct

```python
import numpy as np
import sounddevice as sd
```

```python
FS = 44100                  # sample rate (Hz)
notes = {"E": 82.41, "A": 110.0, "D": 146.83}
```

```python
def power_chord(root, duration=1.0):
    fifth = root * 2**(7/12)
    t = np.arange(int(FS * duration)) / FS    
    tone = np.sin(2*np.pi*root*t) + np.sin(2*np.pi*fifth*t)
    return tone / np.max(np.abs(tone))    
```

```python
if __name__ == "__main__":
    for name, freq in notes.items():
        if freq < 100:
            print(name, "- low chord")
        else:
            print(name, "- chord")
        sd.play(power_chord(freq), FS)
        sd.wait()
```
